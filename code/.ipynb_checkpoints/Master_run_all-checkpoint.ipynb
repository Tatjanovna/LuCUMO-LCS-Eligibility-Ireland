```python
import nbformat
from nbconvert.preprocessors import ExecutePreprocessor
from pathlib import Path
```

0. IMPORTANT: Ensure that the folder “LCS_eligibility_Ireland” is located on your Desktop before running this script.

```python

# 1. BASE DIRECTORY 

BASE = Path.home() / "Desktop" / "LCS_eligibility_Ireland"

CODE_DIR = BASE / "code"


print("Using base directory:", BASE)

```

```python

# 2. NOTEBOOKS TO RUN

notebooks = [
    "1. cleaning_population_projections.ipynb",
    "2. cleaning_smoking_census2022.ipynb",
    "3. transition_rates.ipynb",
    "4. pack_year_distribution_baseline.ipynb",
    "5. Elibability_estimation.ipynb",
]


# 3. EXECUTE NOTEBOOKS

ep = ExecutePreprocessor(timeout=600, kernel_name="python3")

for nb_name in notebooks:
    nb_path = CODE_DIR / nb_name
    print(f"\nRunning notebook: {nb_path}")

    with nb_path.open("r", encoding="utf-8") as f:
        nb = nbformat.read(f, as_version=4)

    # Execute notebook with correct working directory
    ep.preprocess(nb, {"metadata": {"path": str(CODE_DIR)}})

    # Optionally overwrite notebook with executed version
    with nb_path.open("w", encoding="utf-8") as f:
        nbformat.write(nb, f)

print("\nAll notebooks executed successfully.")
```
