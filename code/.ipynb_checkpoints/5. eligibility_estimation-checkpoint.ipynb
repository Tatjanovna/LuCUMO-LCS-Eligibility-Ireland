# 6. Modelling  the base case (2022)

```python
# Eligibility criteria
elig_criteria = {
    "Irish": {"age_min": 55, "age_max": 74, "pack_years": 20, "quit_limit": 15},
    "USPSTF 2013": {"age_min": 55, "age_max": 79, "pack_years": 30, "quit_limit": 15},
    "USPSTF 2021": {"age_min": 50, "age_max": 79, "pack_years": 20, "quit_limit": 15},
    "NELSON": {"age_min": 50, "age_max": 74, "pack_years": 15, "quit_limit": 10},
}

all_results = []

for label, crit in elig_criteria.items():
    threshold_log = np.log(crit["pack_years"])
    quit_suffix = f"excl_{crit['quit_limit']}"

    def age_in_range(ag):
        try:
            low, high = map(int, ag.split("-"))
            return crit["age_min"] <= low and high <= crit["age_max"]
        except:
            return False

    eligible_df = reshaped_data[reshaped_data["age_group"].apply(age_in_range)].copy()
    

    for gender in ["m", "f"]:
        pop_col = f"population_{gender}"
        quitters_col = f"quitters_{gender}"
        smokers_col = f"smokers_{gender}"

        qu_mean_col = f"mean_ln_pack_years_quitters_{quit_suffix}_{gender}"
        qu_std_col = f"std_ln_pack_years_quitters_{quit_suffix}_{gender}"
        sm_mean_col = f"mean_ln_pack_years_smokers_{gender}"
        sm_std_col = f"std_ln_pack_years_smokers_{gender}"



        # Compute proportion eligible (lognormal survival function)
        eligible_df[f"prop_eligible_quitters_{gender}"] = lognorm.sf(
            threshold_log,
            s=eligible_df[qu_std_col],
            scale=np.exp(eligible_df[qu_mean_col])
        )
        eligible_df[f"prop_eligible_smokers_{gender}"] = lognorm.sf(
            threshold_log,
            s=eligible_df[sm_std_col],
            scale=np.exp(eligible_df[sm_mean_col])
        )

        # Multiply by the % of quitters/smokers and population
        eligible_df[f"eligible_quitters_{gender}"] = eligible_df[f"prop_eligible_quitters_{gender}"] * eligible_df[quitters_col]
        eligible_df[f"eligible_smokers_{gender}"] = eligible_df[f"prop_eligible_smokers_{gender}"] * eligible_df[smokers_col]

        eligible_df[f"num_elig_qu_{gender}"] = eligible_df[f"eligible_quitters_{gender}"] * eligible_df[pop_col]
        eligible_df[f"num_elig_sm_{gender}"] = eligible_df[f"eligible_smokers_{gender}"] * eligible_df[pop_col]

    # total = eligible smokers + eligible quitters
    eligible_df["total_eligible_m"] = eligible_df["num_elig_qu_m"] + eligible_df["num_elig_sm_m"]
    eligible_df["total_eligible_f"] = eligible_df["num_elig_qu_f"] + eligible_df["num_elig_sm_f"]
    eligible_df["total_eligible"] = eligible_df["total_eligible_m"] + eligible_df["total_eligible_f"]
    eligible_df["strategy"] = label

    all_results.append(eligible_df)

# Combine results across strategies
combined = pd.concat(all_results, ignore_index=True)

# Define fixed age groups
fixed_age_groups = [f"{a}-{a+4}" for a in range(50, 80, 5)]

# Build region-level summary
summary = []

for strategy in combined["strategy"].unique():
    crit = elig_criteria[strategy]
    age_min, age_max = crit["age_min"], crit["age_max"]

    def group_in_range(ag):
        try:
            lo, hi = map(int, ag.split("-"))
            return lo >= age_min and hi <= age_max
        except:
            return False

    valid_age_groups = [ag for ag in fixed_age_groups if group_in_range(ag)]
    

    for region in combined["region"].unique():
        subset = combined[(combined["strategy"] == strategy) & (combined["region"] == region)]
        total = subset["total_eligible"].sum()
        m_pct = subset["total_eligible_m"].sum() / total * 100 if total > 0 else 0
        f_pct = subset["total_eligible_f"].sum() / total * 100 if total > 0 else 0

        age_group_pcts = {}
        for ag in fixed_age_groups:
            if ag in valid_age_groups:
                age_pct = subset[subset["age_group"] == ag]["total_eligible"].sum() / total * 100 if total > 0 else 0
            else:
                age_pct = 0
            age_group_pcts[f"{ag}(%)"] = age_pct

        row = {
            "Strategy": strategy,
            "Region": region,
            "Estimate": total,
            "Male(%)": m_pct,
            "Female(%)": f_pct,
        }
        row.update(age_group_pcts)
        summary.append(row)

# Convert summary to DataFrame
summary_df = pd.DataFrame(summary).fillna(0)

# Display
summary_df.head(100)


```

```python
# Group across all regions: national-level summary

summary_national = []

for strategy in combined["strategy"].unique():
    crit = elig_criteria[strategy]
    age_min, age_max = crit["age_min"], crit["age_max"]

    def age_in_range(ag):
        try:
            low, high = map(int, ag.split("-"))
            return crit["age_min"] <= low and high <= crit["age_max"]
        except:
            return False

    valid_age_groups = [ag for ag in fixed_age_groups if group_in_range(ag)]

    subset = combined[combined["strategy"] == strategy]
    total = subset["total_eligible"].sum()
    total_m = subset["total_eligible_m"].sum()
    total_f = subset["total_eligible_f"].sum()
    m_pct = total_m / total * 100 if total > 0 else 0
    f_pct = total_f / total * 100 if total > 0 else 0

    age_group_pcts = {}
    for ag in fixed_age_groups:
        if ag in valid_age_groups:
            age_pct = subset[subset["age_group"] == ag]["total_eligible"].sum() / total * 100 if total > 0 else 0
        else:
            age_pct = 0
        age_group_pcts[f"{ag}(%)"] = age_pct

    row = {
        "Strategy (Age & PY)": f"{strategy} (≥{crit['pack_years']} PY, <{crit['quit_limit']}y)",
        "Total Eligible": total,
        "Male Eligible": total_m,
        "Female Eligible": total_f,
        "Male(%)": m_pct,
        "Female(%)": f_pct,
        "50-54(%)": age_group_pcts.get("50-54(%)", 0),
        "55-59(%)": age_group_pcts.get("55-59(%)", 0),
        "60-64(%)": age_group_pcts.get("60-64(%)", 0),
        "65-69(%)": age_group_pcts.get("65-69(%)", 0),
        "70-74(%)": age_group_pcts.get("70-74(%)", 0),
        "75-79(%)": age_group_pcts.get("75-79(%)", 0),
    }

    summary_national.append(row)

# Convert to DataFrame
summary_national_df = pd.DataFrame(summary_national).fillna(0)

# Identify count and percentage columns
count_cols = ["Total Eligible", "Male Eligible", "Female Eligible"]
pct_cols = [col for col in summary_national_df.columns if "(%)" in col]

# Format counts as integers
summary_national_df[count_cols] = summary_national_df[count_cols].astype(int)

# Format percentage columns to 1 or 2 decimals
summary_national_df[pct_cols] = summary_national_df[pct_cols].round(2)

# Display the result
summary_national_df.head(100)

```

```python

# Provided quitting and initiation rate tables
quitting_df = pd.DataFrame({
    'age_group': [
        '15-19', '20-24', '25-29', '30-34', '35-39', '40-44', '45-49',
        '50-54', '55-59', '60-64', '65-69', '70-74', '75-79'
    ],
    'quitting_rate': [
        0.106870, 0.034043, 0.038143, 0.041186, 0.033493, 0.029141,
        0.017778, 0.028736, 0.023098, 0.030075, 0.020080, 0.020478, 0.023121
    ]
})

initiation_df = pd.DataFrame({
    'age_group': [
        '15-19', '20-24', '25-29', '30-34', '35-39', '40-44', '45-49',
        '50-54', '55-59', '60-64', '65-69', '70-74', '75-79'
    ],
    'initiation_rate': [
        0.027778, 0.003831, 0.0, 0.002176, 0.002836, 0.0, 0.000896,
        0.0, 0.000873, 0.0008, 0.0, 0.00075, 0.0
    ]
})

# Merge both rate tables
rates_df = pd.merge(quitting_df, initiation_df, on='age_group')

# Drop rows with missing values
rates_df = rates_df.dropna()


rates_expanded = pd.concat([
    rates_df.assign(gender="Male"),
    rates_df.assign(gender="Female")
], ignore_index=True)

# Function to create transition matrices
def create_transition_matrices(base_rates_df):
    transition_matrices = {}
    for _, row in base_rates_df.iterrows():
        age_group = row["age_group"]
        gender = row["gender"]
        q = row["quitting_rate"]
        i = row["initiation_rate"]

        matrix = np.array([
            [1 - q, q, 0],         # Smokers
            [i, 1 - i, 0],         # Quitters
            [i, 0, 1 - i]          # Never-Smokers
        ])
        transition_matrices[(age_group, gender)] = matrix

    return transition_matrices

# Generate transition matrices
transition_matrices_with_relapse = create_transition_matrices(rates_expanded)

# Also create a version with zero initiation
rates_expanded_zero_initiation = rates_expanded.copy()
rates_expanded_zero_initiation["initiation_rate"] = 0.0

transition_matrices_no_initiation = create_transition_matrices(rates_expanded_zero_initiation)

# Output sample
transition_matrices_with_relapse, transition_matrices_no_initiation
```

```python
def adjust_young_age_group(new_shares, inflow_rate):
    """
    Adjusts the shares of smokers, quitters, and non-smokers for the 15-24 age group and later age groups.
    
    Parameters:
    - new_shares: Array of current shares [smokers, quitters, never-smokers].
    - inflow_rate: Fraction to add to the non-smokers.
    
    Returns:
    - Adjusted shares as a numpy array.
    """
    # Current shares
    smokers, quitters, never_smokers = new_shares

    # Step 1: Increase the share of non-smokers by the inflow rate
    never_smokers += inflow_rate

    # Step 2: Proportionally reduce smokers and quitters
    total_reduction = inflow_rate 
    total_smokers_quitters = smokers + quitters
    if total_smokers_quitters > 0:
        reduction_factor = total_reduction / total_smokers_quitters
        smokers -= smokers * reduction_factor
        quitters -= quitters * reduction_factor
    else:
        # Fallback: Redistribute the reduction proportionally or skip
        smokers = max(smokers - (total_reduction / 2), 0)
        quitters = max(quitters - (total_reduction / 2), 0)

    # Step 3: Renormalize shares to sum to 1
    new_shares = np.array([smokers, quitters, never_smokers])
    new_shares /= new_shares.sum()  # Renormalize to ensure total = 1

    return new_shares

```

```python
import numpy as np

def simulate_one_year_with_inflow(data, transition_matrices, inflow_rate):
    """
    Simulates one year of transitions, including inflows of new non-smokers to multiple cohorts over time.

    Parameters:
    - data: Pandas DataFrame containing the population shares and metadata.
    - transition_matrices: Dictionary of transition matrices for each age-sex group.
    - inflow_rate: Fraction of new non-smokers entering the 15-24 age group.
    - current_year: The simulation year (to determine which cohorts get inflow).
    - start_year: The year the simulation started (to track inflow phase).

    Returns:
    - Updated DataFrame with new population shares.
    """
    updated_data = data.copy()


    for index, row in updated_data.iterrows():
        # Get the transition matrix for this group
        age_group = row["age_group"]
        gender = row["gender"]
        transition_matrix = transition_matrices.get((age_group, gender))

        # Current shares of S, Q, N
        current_shares = np.array([row["smokers"], row["quitters"], row["never_smokers"]])

        print("Share row:", current_shares)
        print("Transition matrix:\n", transition_matrix)

        # Apply the transition matrix
        new_shares = current_shares @ transition_matrix

        # Adjust inflow based on the simulation year
        if age_group == "15-19":
            new_shares = adjust_young_age_group(new_shares, inflow_rate)
        elif age_group == "20-24" and num_years >= 5:
            new_shares = adjust_young_age_group(new_shares, inflow_rate)
        elif age_group == "25-29" and num_years >= 10:
            new_shares = adjust_young_age_group(new_shares, inflow_rate)
        elif age_group == "30-34" and num_years >= 15:
            new_shares = adjust_young_age_group(new_shares, inflow_rate)
        elif age_group == "35-39" and num_years >= 20:
            new_shares = adjust_young_age_group(new_shares, inflow_rate)
        elif age_group == "40-44" and num_years >= 25:
            new_shares = adjust_young_age_group(new_shares, inflow_rate)

        # Update the shares in the dataset
        updated_data.at[index, "smokers"] = new_shares[0]
        updated_data.at[index, "quitters"] = new_shares[1]
        updated_data.at[index, "never_smokers"] = new_shares[2]

    return updated_data
```

```python

# Define a function to simulate transitions over multiple years for shares only
def simulate_shares_over_years(data, transition_matrices, start_year, num_years):
    results = []
    
    # Initialize the current data with the baseline shares
    current_data = data[[ "age_group", "gender", "smokers", "quitters", "never_smokers"]].copy()
    current_data["year"] = start_year

    # Loop over each year
    for year in range(start_year, start_year + num_years):
        # Append the current data to the results
        results.append(current_data.copy())

        # Simulate transitions for the next year
        current_data = simulate_one_year_with_inflow(current_data, transition_matrices, inflow_rate)
        current_data["year"] = year + 1  # Update the year

    # Combine all yearly results into a single dataframe
    return pd.concat(results, ignore_index=True)

# Simulate shares from 2022 
inflow_rate=0.02
start_year = 2022
num_years = 24

projected_shares_base = simulate_shares_over_years(merged_data_ag, transition_matrices_with_relapse, start_year, num_years)

projected_shares_zero = simulate_shares_over_years(merged_data_ag, transition_matrices_no_initiation, start_year, num_years)

# Save to CSV or display

#file_path =  r'C:\Users\tatia\Desktop\projects\lung cancer\eligability\data\eu_barometer_data\projected_shares_only.csv'
#projected_shares_base.to_csv(file_path, index=False)
print(projected_shares_base)

#projected_shares_zero.to_csv(file_path, index=False)
print(projected_shares_zero)
```

```python


# 1. Discounting Function 
def calculate_varying_discounted_value(initial_value, year, start_year, base_rate, rate_increment):
    num_years = year - start_year
    current_value = initial_value
    for t in range(num_years):
        current_rate = base_rate + t * rate_increment
        current_value *= (1 - current_rate)
    return current_value


    """
    Calculate discounted values with a gradually increasing discount rate.
    
    Parameters:
    - initial_value: The starting value to be discounted.
    - year: The current year.
    - start_year: The base year (start of the discounting).
    - base_rate: Initial discount rate (e.g., 1% as 0.01).
    - rate_increment: Yearly increment to the discount rate.
    
    Returns:
    - Discounted value after applying the varying discount rates.
    """

# 2. Apply Discounting 
def apply_discounting_all_variants(df, prefixes, base_rate_mean, rate_increment_mean, base_rate_std, rate_increment_std, start_year):
    for prefix in prefixes:
        mean_col = f'mean_ln_pack_years_{prefix}'
        std_col = f'std_ln_pack_years_{prefix}'
        if mean_col in df.columns and std_col in df.columns:
            df[f'non_discounted_{mean_col}'] = df[mean_col]
            df[f'non_discounted_{std_col}'] = df[std_col]
            df[f'discounted_{mean_col}'] = df.apply(
                lambda row: calculate_varying_discounted_value(row[mean_col], row['year'], start_year, base_rate_mean, rate_increment_mean),
                axis=1
            )
            df[f'discounted_{std_col}'] = df.apply(
                lambda row: calculate_varying_discounted_value(row[std_col], row['year'], start_year, base_rate_std, rate_increment_std),
                axis=1
            )
    return df


```

```python

# 3. Summarize Results 
def group_overlaps_criteria(group, age_min, age_max):
    try:
        start, end = map(int, group.split("-"))
        return not (end < age_min or start > age_max)
    except:
        return False

def summarize_eligibility_totals(df, eligibility_criteria):
    summary = pd.DataFrame()
    summary["year"] = sorted(df["year"].unique())

    for scheme, crit in eligibility_criteria.items():
        suffix = scheme.replace(" ", "_")
        filtered = df[df["age_group"].apply(
            lambda x: group_overlaps_criteria(x, crit["age_min"], crit["age_max"])
        )].copy()

        filtered[f"total_eligible_{suffix}"] = (
            filtered[f"num_elig_sm_{suffix}"] + filtered[f"num_elig_qu_{suffix}"]
        )

        yearly = filtered.groupby("year")[f"total_eligible_{suffix}"].sum().reset_index()
        summary = summary.merge(yearly, on="year", how="left")
        summary.rename(columns={f"total_eligible_{suffix}": suffix}, inplace=True)

    return summary

```

```python
# Base case smoking initiation & zero initiation & 1% attenuation of smoking history

# Prepare both base and zero datasets
df_base = result_with_pack_years_base.copy()
df_zero = result_with_pack_years_zero.copy()
df_base_no_att = result_with_pack_years_base.copy()



# Eligibility Processing 
def process_eligibility_aligned_with_census(df, scheme_name, criteria, quit_pct_dict, discounted=True):
    prefix = "discounted_" if discounted else ""
    suffix = scheme_name.replace(" ", "_")
    quit_suffix = "excl_10" if criteria["quit_limit"] == 10 else "excl_15"

    mean_quit_col = f"{prefix}mean_ln_pack_years_quitters_{quit_suffix}"
    std_quit_col = f"{prefix}std_ln_pack_years_quitters_{quit_suffix}"
    
    df[f"{prefix}mean_ln_pack_years_smokers"] = df[f"{prefix}mean_ln_pack_years_smokers"].replace(0, np.nan)
    df[mean_quit_col] = df[mean_quit_col].replace(0, np.nan)
    
    df = df.dropna(subset=[
        f"{prefix}mean_ln_pack_years_smokers",
        f"{prefix}std_ln_pack_years_smokers",
        mean_quit_col,
        std_quit_col
    ])

    threshold_log = np.log(criteria["pack_years"])
    
    df[f"prop_eligible_smokers_{suffix}"] = lognorm.sf(
        threshold_log,
        s=df[f"{prefix}std_ln_pack_years_smokers"],
        scale=np.exp(df[f"{prefix}mean_ln_pack_years_smokers"])
    )
    df[f"prop_eligible_quitters_{suffix}"] = lognorm.sf(
        threshold_log,
        s=df[std_quit_col],
        scale=np.exp(df[mean_quit_col])
    )

    adjusted_quitters = []
    for idx, row in df.iterrows():
        age_grp = row["age_group"]
        quit_fraction = quit_pct_dict.get(age_grp, {}).get(f"<{criteria['quit_limit']}", 0)
        val = row[f"prop_eligible_quitters_{suffix}"] * quit_fraction * row["quitters"] * row["population"]
        adjusted_quitters.append(val)
    df[f"num_elig_qu_{suffix}"] = adjusted_quitters

    df[f"num_elig_sm_{suffix}"] = (
        df[f"prop_eligible_smokers_{suffix}"] * df["smokers"] * df["population"]
    )

    return df

# Execute for Base and Zero Scenarios 
base_rate_mean = 0.01
rate_increment_mean = 0.001
base_rate_std = 0.001
rate_increment_std = 0.0001
start_year = 2023
all_prefixes = ['smokers', 'quitters_all', 'quitters_excl_10', 'quitters_excl_15']

df_base = apply_discounting_all_variants(
    df_base, all_prefixes,
    base_rate_mean, rate_increment_mean,
    base_rate_std, rate_increment_std,
    start_year
)

df_zero = apply_discounting_all_variants(
    df_zero, all_prefixes,
    base_rate_mean, rate_increment_mean,
    base_rate_std, rate_increment_std,
    start_year
)

for scheme, crit in eligibility_criteria.items():
    df_base = process_eligibility_aligned_with_census(df_base, scheme, crit, quit_pct_dict=quit_pct, discounted=True)
    df_zero = process_eligibility_aligned_with_census(df_zero, scheme, crit, quit_pct_dict=quit_pct, discounted=True)
    df_base_no_att = process_eligibility_aligned_with_census(df_base_no_att, scheme, crit, quit_pct_dict=quit_pct, discounted=False)

summary_base = summarize_eligibility_totals(df_base, eligibility_criteria)
summary_zero = summarize_eligibility_totals(df_zero, eligibility_criteria)
summary_no_att = summarize_eligibility_totals(df_base_no_att, eligibility_criteria)

summary_base.head().round(0).astype(int)
```

```python
import matplotlib.pyplot as plt

# Create subplots for each eligibility criteria
criteria_names = ["Irish", "NELSON", "USPSTF_2013", "USPSTF_2021"]
scenario_dfs = {
    "Basecase + 1% Attenuation": summary_base,
    "No initiation + 1% Attenuation": summary_zero,
    "Basecase + 3% Attenuation" : summary_base_high,
    "No initiation + 3% Attenuation": summary_zero_high,
    "Basecase + No Attenuation": summary_no_att
}

fig, axs = plt.subplots(nrows=2, ncols=2, figsize=(16, 10), sharex=True, sharey=True)
axs = axs.flatten()

# Define style mappings
scenario_colors = {
    "Basecase + 1% Attenuation": "blue",
    "No initiation + 1% Attenuation": "green",
    "Basecase + 3% Attenuation": "blue",
    "No initiation + 3% Attenuation": "green",
    "Basecase + No Attenuation": "orange"
}

scenario_styles = {
    "Basecase + 1% Attenuation": "-",
    "No initiation + 1% Attenuation": "-",
    "Basecase + 3% Attenuation": "--",
    "No initiation + 3% Attenuation": "--",
    "Basecase + No Attenuation": "-"
}

# Plot each criteria
for idx, criteria in enumerate(criteria_names):
    ax = axs[idx]
    for scenario, df in scenario_dfs.items():
        if criteria in df.columns:
            ax.plot(df["year"], df[criteria], label=scenario,
                    color=scenario_colors[scenario],
                    linestyle=scenario_styles[scenario])
    ax.set_title(f"{criteria} Eligibility")
    ax.set_xlabel("Year")
    ax.set_ylabel("Eligible Population")
    ax.grid(True)

# Shared legend
handles, labels = axs[0].get_legend_handles_labels()
fig.legend(handles, labels, loc="upper center", ncol=5, bbox_to_anchor=(0.5, 1.05))

plt.tight_layout(rect=[0, 0, 1, 0.95])
file_path =  OUT_DIR / "eligability_projections.png"
plt.savefig(file_path, dpi=300, bbox_inches='tight')
plt.show()

```
